### MISSING VALUE ANALYSIS

```python
#To check for missing values
uber_df.isnull()
```

```python
#To check whether at least one missing value exists in each column.
uber_df.isnull().any()
```

```python
#To count the number of missing (NaN or None) values in each column of a DataFrame.
missing_count = uber_df.isnull().sum().sort_values(ascending = False)

#To calculate missing value percentage.
missing_percentage = (missing_count / len(uber_df)*100).round(2)

#Missing value summary
missing_summary = pd.DataFrame({'Missing Count': missing_count, 'Missing %': missing_percentage})

#Print it
print(missing_summary)
```

```python
#Visualizing missing values

plt.figure(figsize=(10,6))
missing_nonzero = missing_count[missing_count > 0]
sns.barplot(x=missing_nonzero .values,y=missing_nonzero .index, palette ='rocket')
plt.xlabel("Number of missing values")
plt.ylabel("Name of the columns")
plt.title("Barplot for visualizing missing values")
plt.tight_layout()
plt.show()
```

```python
#Handling Missing Values for reason columns.

Reason_columns = [
    'Incomplete Rides Reason',
    'Reason for cancelling by Customer',
    'Driver Cancellation Reason'
]

for col in Reason_columns:
    uber_df[col] = uber_df[col].fillna('Not Applicable')
```

```python
numeric_columns = [
    'waiting time', 
    'Avg CTAT', 
    'Booking Value', 
    'Ride Distance', 
    'Driver Ratings', 
    'Customer Rating'
]

for num in numeric_columns:
    uber_df[num] = pd.to_numeric(uber_df[num], errors='coerce')
    
    uber_df[num] = uber_df[num].fillna(
        uber_df[num].median()
    )
```

```python
# Categorical columns with  missing entries fill with mode (most frequent entries).

categorical_cols = ['Vehicle Type', 
                    'Pickup Location', 
                    'Drop Location',
                    'Payment Method',
                    'Customer ID']

for col in categorical_cols:
    if uber_df[col].isnull().sum() > 0:
        uber_df[col] = uber_df[col].fillna(uber_df[col].mode()[0])
```

### DATA TIME ANALYSIS

```python
#To Display all the column names
uber_df.columns
```

```python
#Inspect date values
uber_df['Date'].head()

#To convert date to pandas datetime format
uber_df['Date'] = pd.to_datetime(
    uber_df['Date'],
    errors='coerce'
)

#To extract year from date
uber_df['Year'] = uber_df['Date'].dt.year
```

```python
#To extract month from date
uber_df['Month'] = uber_df['Date'].dt.month
```

```python
#To extract Month names
uber_df['Month_Name'] = uber_df['Date'].dt.month_name()
```

```python
#To extract day of the month
uber_df['Day'] = uber_df['Date'].dt.day
```

```python
#To extract day of week
uber_df['Day_of_week'] = uber_df['Date'].dt.day_name()
```

```python
#To create a column for day type as weekdays / weekend
uber_df['Is_Weekend'] = np.where(
    uber_df['Date'].dt.dayofweek >= 5,
    'Yes',
    'No'
)
```

```python
#To check the date categories 
uber_df[['Year','Month','Month_Name','Day','Day_of_week','Is_Weekend']].head()
```

```python
#TO check data type of time
uber_df['Time'].dtype
```

```python
#To convert time to proper pandas datetime format
uber_df['Time'] = pd.to_datetime(
    uber_df['Time'],
    format='%H:%M:%S',
    errors='coerce'
)
```

```python
#To extract hour
uber_df['Hour'] = uber_df['Time'].dt.hour
```

```python
#To Create Day-Night Category 
def day_night(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour < 17:
        return 'Afternoon'
    elif 17 <= hour < 21:
        return 'Evening'
    else:
        return 'Night'

uber_df['Day-Night'] = uber_df['Hour'].apply(day_night)
```

```python
#Creating a "Peak Hour" category
def peak_hour(hour):
    if 7 <= hour <= 10 or 17 <= hour <= 20:
        return 'Peak Hour'
    else:
        return 'Non-Peak Hour'

uber_df['Peak_Hour'] = uber_df['Hour'].apply(peak_hour)
```
